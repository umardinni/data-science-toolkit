```python
import numpy as np
array_2x2 = np.array([
    [1,2],
    [3,4]
])
print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
random3x4 = np.random.rand(3,4)
print("Random 3x4 array")
print(random3x4)
```

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))
print("Array of Ones:")
print(ones_array)
print("Array of Zeros:")
print(zeros_array)
```

```python
arr = np.random.rand(6, 3)


arr_transposed = arr.T

print("Original Array (6x3):\n", arr)
print("\nTransposed Array (3x6):\n", arr_transposed)
```

```python
n, p, m = 3, 4, 2

A = np.random.rand(n, p)
B = np.random.rand(p, m)

C = np.dot(A, B)

print("Array A (size {}x{}):\n{}\n".format(n, p, A))
print("Array B (size {}x{}):\n{}\n".format(p, m, B))
print("Dot Product (size {}x{}):\n{}".format(n, m, C))
```
